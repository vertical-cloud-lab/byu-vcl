```python
import re
patterns = {
'byu_coi': ['Related Party','negotiat','one day','four days','compensat','secondary employment','approve','business'],
'byu_fcoi': ['management plan','AAVP','five business days','retrospective','30 days','training','chair','dean'],
'byu_ip': ['45','ten percent','10 percent','$3','3,000,000','board','consultant','sponsored research','line officer','name of the university'],
'stanford_best': ['student','management','board','publication','principal investigator','facility','equipment','consult'],
'stanford_students': ['student','thesis','dissertation','advisor','employment','internship','publication','choice'],
'iowa_plan': ['management','monitor','student','review','duration','appeal','acknowledge'],
'iowa_startup': ['student','research','management','equity','SBIR','STTR'],
'nih': ['five business days','retrospective','120 days','public','monitor','human subjects']}
for key, pats in patterns.items():
    text=pages[key]
    print('\n===== ',key,' =====')
    lines=text.splitlines()
    for p in pats:
        hits=[(i,l.strip()) for i,l in enumerate(lines) if p.lower() in l.lower()]
        print('\n--',p,'--')
        for i,l in hits[:6]: print(i, l[:500])
```

```python
def contexts(text, terms, radius=3, maxhits=10):
    ls=text.splitlines()
    for term in terms:
        print(f'\n### {term}')
        hits=[i for i,l in enumerate(ls) if term.lower() in l.lower()]
        for i in hits[:maxhits]:
            print('\n'.join(f'{j}: {ls[j]}' for j in range(max(0,i-radius),min(len(ls),i+radius+1))))
            print('---')
contexts(pages['byu_fcoi'], ['develop a management plan','30 days','five business days','retrospective review','training','Associate Academic Vice President for Research','dean'])
contexts(pages['byu_ip'], ['45 percent','10 percent','$3 million','voting member','paid consultant','line officers','sponsored research','advertising'], radius=2)
```

```python
# Print concise exact extracts from BYU sources and peer guidance for final verification.
def show_matches(key, regexes, before=2, after=4):
    lines=pages[key].splitlines()
    print(f'\n===== {key} =====')
    used=set()
    for rx in regexes:
        inds=[i for i,l in enumerate(lines) if re.search(rx,l,re.I)]
        print(f'\nPATTERN {rx}: {len(inds)} hits')
        for i in inds[:8]:
            span=range(max(0,i-before),min(len(lines),i+after+1))
            sig=(min(span),max(span))
            if sig in used: continue
            used.add(sig)
            print('\n'.join(f'{j}: {lines[j]}' for j in span)); print('---')
show_matches('byu_fcoi',[r'Management Plan',r'public.*five business days',r'training',r'monitor',r'AAVP.*R',r'department chair and college dean'],1,3)
show_matches('byu_ip',[r'45\s*percent',r'10\s*percent|ten percent',r'3 million|3,000,000',r'voting',r'line officer',r'paid consultant'],1,3)
show_matches('stanford_best',[r'student',r'management',r'publication',r'principal investigator|PI',r'facilit',r'equipment'],1,2)
show_matches('stanford_students',[r'advisor',r'thesis|dissertation',r'employ',r'choice|voluntary',r'non.?retali',r'publish'],1,2)
show_matches('iowa_plan',[r'monitor',r'review',r'student',r'acknowledge',r'duration|termination',r'dispute|appeal'],1,2)
```
