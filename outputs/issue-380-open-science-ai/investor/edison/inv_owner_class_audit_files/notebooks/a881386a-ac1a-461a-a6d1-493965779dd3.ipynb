```python
from e14c.web import fetch_parsed_page, save_citation
import asyncio, re, json, datetime
plan = ['Classify accounts from rules and supplied identity metadata; ignore ranks and stars.', 'Check uncertain ownership, commercial product links and foundation versus fiscal-host governance using primary websites or READMEs.', 'Write the requested seven-column CSV; validate order, completeness and basis lengths; compute counts and explain all judgment calls.']
print('\n'.join(plan))
print('Access date:',datetime.date.today())
txt=await fetch_parsed_page('https://anoma.ly')
print(txt[:16000])
```

```python
evidence={}
citation_paths=set()
def clean_content(text):
    # Exclude incidental outcome-bearing lines; they are never evidence for this audit.
    excluded=re.compile(r'\b(funding|fundrais\w*|valuation|revenue|acquisition price|shutdown|shut down|IPO|raised \$|raised €|series [a-f]\b|seed round)\b',re.I)
    return '\n'.join(line for line in str(text).splitlines() if not excluded.search(line))
def cite_page(key,url,text,author):
    headings=re.findall(r'^#+\s+(.+)$',text,re.M)
    first=next((s.strip() for s in text.splitlines() if s.strip()),key)
    title=(headings[0] if headings else first)[:220]
    title=re.sub(r'!?\[([^]]*)\]\([^)]*\)',r'\1',title)
    def esc(s):
        return str(s).replace('\\','').replace('{','').replace('}','').replace('&',r'\&').replace('_',r'\_').replace('%',r'\%').replace('#',r'\#')
    ck=re.sub('[^A-Za-z0-9]','',key)+'2026source'+str(len(evidence)+1)
    bib='@online{'+ck+',\n  author = {{'+esc(author)+'}},\n  title = {'+esc(title)+'},\n  url = {'+url+'},\n  urldate = {'+str(datetime.date.today())+'}\n}\n'
    citation_paths.add(str(save_citation(bib)))
    return ck
async def check(key,url,author=None,limit=10500):
    try:
        raw=await fetch_parsed_page(url)
        safe=clean_content(raw)
        ck=cite_page(key,url,safe,author or key)
        evidence.setdefault(key,[]).append({'url':url,'text':safe,'citation_key':ck})
        return key,safe[:limit]
    except Exception as e:
        return key,'FETCH FAILED: '+str(e)[:200]
# Preserve and cite first page already checked.
safe=clean_content(txt)
ck=cite_page('anomalyco','https://anoma.ly',safe,'Anomaly')
evidence['anomalyco']=[{'url':'https://anoma.ly','text':safe,'citation_key':ck}]
queries=[('NousResearch','https://nousresearch.com'),('anomalyco','https://opencode.ai/zen'),('FuelLabs','https://fuel.network'),('ultraworkers','https://github.com/ultraworkers/claw-code'),('open-webui','https://openwebui.com'),('Graphify-Labs','https://graphify.com/'),('OpenCut-app','https://www.opencut.app'),('oven-sh','https://bun.com/')]
res=await asyncio.gather(*(check(*q,limit=10500) for q in queries))
for k,t in res: print('\n========',k,'========\n',t)
```
