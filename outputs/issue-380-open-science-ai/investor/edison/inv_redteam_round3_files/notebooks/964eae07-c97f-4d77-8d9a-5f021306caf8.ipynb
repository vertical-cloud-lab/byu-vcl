```python
r=data['h1_returns']; l=data['cohort_listings_verified']
o=r[(r.cohort=='open_source') & (r.company!='JFrog')].copy();c=r[r.cohort=='closed'].copy()
for name,d in [('open',o),('closed',c)]:
 print(name,'n',len(d),'relative wealth median',(d.multiple_from_offer/d.qqq_multiple).median(),'excess QQQ',d.excess_vs_qqq_offer.median(),'lost',d.lost_money_offer.sum(),'first closes',d.first_day_close.notna().sum())
 print('2021–22',d[d.listing_date.str[:4].isin(['2021','2022'])][['company','listing_date']].to_string(index=False))
print('\nKey companies in return file:')
print(r[r.company.isin(['JFrog','SUSE','New Relic','MongoDB','Talend','MuleSoft','MariaDB','HashiCorp','SolarWinds'])].to_string(index=False))
print('\nVerified listing details:')
print(l[l.company.isin(['SUSE','New Relic','MongoDB','Talend','MuleSoft','MariaDB','HashiCorp','SolarWinds'])][['company','offer_price','first_day_close','exit_price_per_share','exit_consideration_detail','notes']].to_string(index=False))
print('Verified count',len(l))
```

```python
import numpy as np
import statsmodels.formula.api as smf
primary=pd.concat([o,c],ignore_index=True)
primary['log_pme']=np.log(primary.multiple_from_offer/primary.qqq_multiple)
primary['is_open']=(primary.cohort=='open_source').astype(int)
primary['listing_year']=primary.listing_date.str[:4]
model=smf.ols('log_pme ~ is_open + C(listing_year)',data=primary)
for cov in ['nonrobust','HC0','HC1','HC2','HC3']:
 fit=model.fit(cov_type=cov)
 print(cov,'ratio',np.exp(fit.params['is_open']),'CI',np.exp(fit.conf_int().loc['is_open']).to_list(),'p',fit.pvalues['is_open'])
print('\nPrimary reproduced annual gaps:')
for col in ['excess_vs_qqq_offer','excess_vs_qqq_first','excess_vs_igv_offer']:
 print(col,(o[col].median()-c[col].median())*100)
print('\nCohort vs verified offer values:')
merged=primary.merge(l[['company','offer_price','first_day_close']],on='company',suffixes=('_returns','_listings'),how='left',indicator=True)
print(merged.loc[~np.isclose(merged.offer_price_returns,merged.offer_price_listings,equal_nan=True),['company','offer_price_returns','offer_price_listings']].to_string(index=False))
print('\nPrimary cohort names')
print(primary[['company','listing_type']].to_string(index=False))
```

```python
fit_t=model.fit(cov_type='HC3',use_t=True)
print('HC3, small-sample t interval:',np.exp(fit_t.conf_int().loc['is_open']).tolist(),'p:',fit_t.pvalues['is_open'])
print('\nVerified names absent in primary exact-name join:')
print(l[['company','offer_price','first_day_close']].to_string(index=False))
print('\nFraming edit counts:',data['round2_framing_edits'].direction.value_counts().to_dict())
print('\nOld phrases remaining verbatim:')
text=' '.join(Path('INVESTOR_MEMO.md').read_text().split())
for _,row in data['round2_framing_edits'].iterrows():
 if ' '.join(row.memo_quote.split()) in text:print(row.memo_quote)
```
